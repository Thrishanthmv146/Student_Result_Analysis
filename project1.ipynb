```python
import pandas as pd
import numpy as np
students={"_Student_Id":[101,102,103,104,105,106,107,108,109,110,111,112,113,114,115],
          "_Name":["Thrish","Harsha","Ravi","Tejas","Yadhav","Suresh","Raju","Preetam","Maju","Pravin","Chiranth","Yash","Virat","Rohith","Sanju"],

          "_Sub:Python":[57,79,43,66,35,71,87,45,70,32,78,98,89,93,44],
           "_Sub:Data Science":[67,89,23,76,65,21,57,95,80,22,88,78,49,90,50],
           "_Sub:Agentic Ai":[47,59,33,56,55,31,97,75,90,62,88,98,59,80,40],
           "_Sub:Maths":[67,89,23,56,35,41,97,45,90,22,78,88,89,70,60],
          "_Attendence":["78%","67%","89%","95%","78%","50%","60%","45%","56%","63%","77%","33%","55%","70%","45%"]}
df=pd.DataFrame(students)
df.to_csv("students.csv",index=False)

print("                                 ============= STUDENT DATA ===============\n")
print(df,"\n")

def total_marks(marks):
  df["_Total_Marks"]=(df["_Sub:Python"]+df["_Sub:Data Science"]+df["_Sub:Agentic Ai"]+df["_Sub:Maths"])
  return df
total_marks(df)

def average_marks(average):
  df["_Average"]=(df["_Total_Marks"]/4)
  return df
average_marks(df)


for index,row in df.iterrows():
  if row["_Average"]>=90:
   df.loc[index, "_Grade"]="A+"
  elif row["_Average"]>=80 and row["_Average"]<90:
   df.loc[index,"_Grade"]="A"
  elif row["_Average"]>=60 and row["_Average"]<80:
   df.loc[index,"_Grade"]="B"
  elif row["_Average"]>=40 and row["_Average"]<60:
   df.loc[index, "_Grade"]="C"
  else:
   df.loc[index, "_Grade"]="Fail"

for index,row in df.iterrows():
  if row["_Average"]>39:
    df.loc[index, "_PASS/FAIL"]="Pass"
  else:
    df.loc[index, "_PASS/FAIL"]="Fail"

Totalmarks=pd.DataFrame(df[["_Student_Id","_Name","_Total_Marks","_Average","_Grade","_PASS/FAIL"]])
print("                                    ============== STUDENT PERFORMANCE DATA =============== \n")
print(Totalmarks,"\n")
subjects=["_Sub:Python","_Sub:Data Science","_Sub:Agentic Ai","_Sub:Maths"]

print("       ///===/HGHEST/=========/AVERAGE/==========/LOWEST/============PERFORMARS=====//// \n ")

# Convert dictionary into DataFrame
df = pd.DataFrame(students)

# Extract subject columns
subjects = df.filter(like="_Sub:")

# Calculate total and average marks
df["_Total_Marks"] = subjects.sum(axis=1)
df["_Average"] = subjects.mean(axis=1)

#  highest, average, and lowest performers
highest = df["_Total_Marks"].max()
lowest = df["_Total_Marks"].min()
class_average = df["_Total_Marks"].mean()

# Select the student closest to the class average
average_performer = (df["_Total_Marks"] - class_average).abs().idxmin()

# Store the selected students in another DataFrame
result_df = pd.concat([
    df[df["_Total_Marks"] == highest],
    df.loc[[average_performer]],
    df[df["_Total_Marks"] == lowest]
])

# Keep only the ID, name, total, and average columns
result_df = result_df[
    ["_Student_Id", "_Name", "_Total_Marks", "_Average"]
].copy()

# Label each performer's category
result_df.insert(
    0, "Performance",
    ["Highest"] * len(df[df["_Total_Marks"] == highest])
    + ["Average"]
    + ["Lowest"] * len(df[df["_Total_Marks"] == lowest])
)

print(result_df,"\n")

print("                            /=========/TOP PERFORMING STUDENTS/====================== \n")
mean_average=df["_Average"].mean()

#  DataFrame to get students above mean average
Top_Results = df[df["_Average"] >= mean_average].copy()

# Select specific columns with double brackets
Top_Results = Top_Results[["_Student_Id", "_Name", "_Average"]].copy()

# Sort by average score descending
Top_Results = Top_Results.sort_values(by="_Average", ascending=False)

Top_Results.insert(
    0, "Performance",
    ["Top"] * len(Top_Results))
print(Top_Results)
```
